```python
import numpy as np

X = np.random.rand(
    100,
    32
).astype(
    np.float32
)

from molaug_qcbm.baselines.vae import MolecularVAE

vae = MolecularVAE(
    epochs=1,
    batch_size=16
)

vae.fit(X)

print(
    "VAE OK",
    vae.generate(5).shape
)
```

```python
import numpy as np

from molaug_qcbm.baselines.gan import (
    MolecularGAN
)

X = np.random.rand(
    100,
    32
).astype(
    np.float32
)

gan = MolecularGAN(
    latent_dim=16,
    epochs=1,
    batch_size=16
)

gan.fit(X)

generated = gan.generate(
    n_samples=5
)

print(
    "Generated Shape:",
    generated.shape
)

print(
    gan.summary()
)
```

```python
import numpy as np

from molaug_qcbm.baselines.qvae import (
    QuantumVAE
)

X = np.random.rand(
    50,
    32
).astype(
    np.float32
)

qvae = QuantumVAE(
    latent_dim=8,
    n_qubits=5,
    epochs=1,
    batch_size=8
)

qvae.fit(X)

generated = qvae.generate(
    n_samples=5
)

print(
    "Generated Shape:",
    generated.shape
)

print(
    qvae.summary()
)
```

```python
import numpy as np

from molaug_qcbm.baselines.qgan import (
    QuantumGAN
)

X = np.random.rand(
    50,
    32
).astype(
    np.float32
)

qgan = QuantumGAN(
    n_qubits=5,
    n_layers=2,
    epochs=1,
    batch_size=8
)

qgan.fit(X)

generated = qgan.generate(
    n_samples=5
)

print(
    "Generated Shape:",
    generated.shape
)

print(
    qgan.summary()
)
```

```python
print("=" * 80)
print("BASELINE BENCHMARK SANITY TEST")
print("=" * 80)

print("GAN")
print(gan.summary())

print("\nQVAE")
print(qvae.summary())

print("\nQGAN")
print(qgan.summary())

print("\nGAN Samples")
print(gan.generate(3).shape)

print("\nQVAE Samples")
print(qvae.generate(3).shape)

print("\nQGAN Samples")
print(qgan.generate(3).shape)

print("\nALL BASELINE TESTS PASSED")
```
